```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print("All libraries imported successfully!")
```

```python
import pandas as pd

df = pd.read_csv("retail_store_sales.csv")

print("Dataset loaded successfully!")
```

```python
# First 5 rows
display(df.head())

# Number of rows and columns
print("Dataset Shape:", df.shape)

# Column names
print("Columns:", df.columns.tolist())

# Data types
print(df.dtypes)

# Missing values
print(df.isnull().sum())

# Basic statistics
display(df.describe(include="all"))
```

```python
# 1. Select specific columns
display(df[['Transaction ID', 'Category', 'Quantity', 'Price Per Unit']].head())

# 2. Filter data
filtered_data = df[df['Quantity'] > 2]
display(filtered_data.head())

# 3. Sort data
sorted_data = df.sort_values(by='Price Per Unit', ascending=False)
display(sorted_data.head())

# 4. Group data
category_sales = df.groupby('Category')['Quantity'].sum()
display(category_sales)

# 5. Transform data
df['Total Sales'] = df['Quantity'] * df['Price Per Unit']

display(df.head())
```

```python
# 1. Check missing values
print("Missing Values:")
print(df.isnull().sum())

# 2. Check duplicate records
print("Duplicate Rows:", df.duplicated().sum())

# 3. Check data types
print("\nData Types:")
print(df.dtypes)

# 4. Remove duplicate rows
df = df.drop_duplicates()

# 5. Fill missing values
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].fillna('Unknown')

for col in df.select_dtypes(include='number').columns:
    df[col] = df[col].fillna(df[col].median())

# 6. Check cleaning result
print("\nMissing Values After Cleaning:")
print(df.isnull().sum())

print("\nDuplicate Rows After Cleaning:", df.duplicated().sum())
```

```python
# Convert Transaction Date into datetime
df['Transaction Date'] = pd.to_datetime(
    df['Transaction Date'],
    errors='coerce'
)

# Check Discount Applied values
print("Discount Values:")
print(df['Discount Applied'].value_counts())

# Check updated data types
print("\nUpdated Data Types:")
print(df.dtypes)

# Check invalid dates
print("\nInvalid Dates:", df['Transaction Date'].isnull().sum())
```

```python
df.to_csv("cleaned_retail_sales.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
# Extract month from transaction date
df['Month'] = df['Transaction Date'].dt.to_period('M').astype(str)

# Group sales by category
category_sales = df.groupby('Category')['Total Sales'].sum()

# Group sales by month
monthly_sales = df.groupby('Month')['Total Sales'].sum()

# Group quantity by category
category_quantity = df.groupby('Category')['Quantity'].sum()

print("Data prepared for visualization!")
```

```python
plt.figure(figsize=(9,5))
category_sales.plot(kind='bar')
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,5))
plt.hist(df['Total Sales'], bins=30)
plt.title('Distribution of Total Sales')
plt.xlabel('Total Sales')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(12,5))
monthly_sales.plot(kind='line', marker='o')
plt.title('Monthly Sales Trend')
plt.xlabel('Month')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.grid()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(9,5))
plt.scatter(df['Quantity'], df['Total Spent'], alpha=0.5)
plt.title('Quantity vs Total Spent')
plt.xlabel('Quantity')
plt.ylabel('Total Spent')
plt.show()
```

```python
plt.figure(figsize=(9,5))
sns.boxplot(x=df['Total Sales'])
plt.title('Box Plot of Total Sales')
plt.xlabel('Total Sales')
plt.show()
```

```python
# 1. Overall Sales
total_sales = df['Total Sales'].sum()

# 2. Average Sales
average_sales = df['Total Sales'].mean()

# 3. Total Transactions
total_transactions = df['Transaction ID'].nunique()

# 4. Best Selling Category
best_category = df.groupby('Category')['Quantity'].sum().idxmax()

# 5. Highest Sales Category
highest_sales_category = df.groupby('Category')['Total Sales'].sum().idxmax()

# 6. Average Quantity
average_quantity = df['Quantity'].mean()

# Display Results
print("----- RETAIL SALES ANALYSIS -----")
print("Total Sales:", round(total_sales, 2))
print("Average Sales:", round(average_sales, 2))
print("Total Transactions:", total_transactions)
print("Best Selling Category:", best_category)
print("Highest Sales Category:", highest_sales_category)
print("Average Quantity:", round(average_quantity, 2))
```

```python
# Descriptive Statistics
display(df[['Quantity', 'Price Per Unit', 'Total Spent', 'Total Sales']].describe())

# Correlation Analysis
correlation = df[['Quantity', 'Price Per Unit', 'Total Spent', 'Total Sales']].corr()

display(correlation)

# Correlation Heatmap
plt.figure(figsize=(8,5))
sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Retail Sales Correlation Heatmap')
plt.show()
```
